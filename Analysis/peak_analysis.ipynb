```python
int_tot_x_off = 0.
for i in range(1, 11):
    img =Image(f'../raw_data/2025-12-16/img/odt=0.0mW_2_{i:02d}.fits.gz')
    img.select_roi(0, 450, 200, 400)
    int_tot_x_off += img.row_sum

int_tot_x_off /= 10

plt.plot(int_tot_x_off)
plt.show()
```

```python
max_vals = []
shim_x_vals = np.arange(-1100., -300., 50.)

for val in shim_x_vals:
    int_x = 0.
    for i in range(1, 4):
        img =Image(f'../raw_data/2025-12-16/img/shimX={val:.0f}mA_odt=1300mW_{i:02d}.fits.gz')
        img.select_roi(0, 450, 200, 400)
        int_x += img.row_sum
    
    int_x /= 3
    
    max_val = np.max(int_x - int_tot_x_off)
    max_vals.append(max_val)

plt.plot(shim_x_vals, max_vals)
plt.show()
```
